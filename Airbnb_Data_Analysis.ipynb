```python
## Top 10 Most Expensive Airbnb Listings
```

```python
top_expensive = ( df[["name" , "neighbourhood_group" , "room_type" , "price"]]
                .sort_values(by="price" , ascending=False)
                .head(10))
top_expensive
```

```python
plt.figure(figsize=(14,6))
sns.barplot(
    data=top_expensive,
    x = "price",
    y = "name",
    palette="rainbow"
)
plt.title("Top 10 Most Expensive Airbnb Listings")
plt.xlabel("Price")
plt.ylabel("Listing")

plt.show()
```

```python
avg_price = (
    df.groupby("neighbourhood_group")["price"]
      .mean()
      .sort_values(ascending=False)
      .reset_index()
)

avg_price
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    data=avg_price,
    x="neighbourhood_group",
    y="price",
    palette="viridis"
)

plt.title("Average Airbnb Price by Borough")

plt.show()
```

```python
top_neighbourhood = (
    df.groupby("neighbourhood")["price"]
      .mean()
      .sort_values(ascending=False)
      .head(15)
      .reset_index()
)
```

```python
plt.figure(figsize=(12,7))

sns.barplot(
    data=top_neighbourhood,
    x="price",
    y="neighbourhood",
    palette="rocket"
)

plt.title("Top 15 Most Expensive Neighbourhoods")

plt.show()
```

```python
fig = px.scatter_map(
    df.sample(3000),
    lat="latitude",
    lon="longitude",
    color="room_type",
    size="price",
    hover_name="name",
    hover_data=[
        "neighbourhood_group",
        "price"
    ],
    zoom=10,
    height=700,
    title="Airbnb Listings in NYC"
)

fig.show()
```

```python
### 📊 Executive Summary

## Key Findings

##- The dataset contains nearly 49,000 Airbnb listings across New York City.
- Manhattan and Brooklyn dominate the Airbnb market in terms of listings.
- Entire Home/Apt is the most frequently listed room type.
- Listing prices are heavily right-skewed, with a small number of luxury properties driving the highest prices.
- Minimum-night requirements are generally low, making most listings suitable for short stays.
- Professional hosts manage a significant number of properties.
- Premium neighbourhoods command substantially higher average prices.
- Interactive geographic visualization highlights dense Airbnb activity in Manhattan and Brooklyn.
- Correlation analysis indicates that pricing depends on multiple factors rather than a single numerical variable.
- The project demonstrates the complete data analysis workflow, from data cleaning to business insights and visualization.
```
